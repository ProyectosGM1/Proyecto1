```python
import networkx as nx
import matplotlib.pyplot as plt

# Crear el grafo dirigido
G = nx.DiGraph()

# Agregar las relaciones de la tabla (padre -> hija -> nieta)
# Los nodos son los nombres y las aristas representan las relaciones padre -> hijo

# Nivel 1 (Padres)
G.add_edges_from([("Juan", "Ana"), ("Juan", "Luis"),
                 ("Maria", "Ana"), ("Maria", "Luis")])

# Nivel 2 (Hijas -> Nietas)
G.add_edges_from([("Ana", "Carla"), ("Luis", "Sara"),
                 ("Ana", "Sara"), ("Luis", "Carla")])

# Visualizar el grafo
plt.figure(figsize=(8, 6))
pos = nx.spring_layout(G)  # Layout para los nodos
nx.draw(G, pos, with_labels=True, node_size=3000, node_color="lightblue", font_size=10, font_weight="bold", arrowsize=20)
plt.title("Árbol de Relaciones Familiares")
plt.show()
```
